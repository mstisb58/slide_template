```python
import os
import sys
sys.path.insert(0, os.path.abspath('pyslides'))

from pyslides import Deck
import plotly.express as px
import plotly.graph_objects as go

deck = Deck(
    title="NIR-HSIを用いた毛髪診断システムの開発",
    theme="kracie",
    style="assets/css/theme-custom.css" # 実際に使用するCSSの相対パスを指定
)
```

```python
# スライド1: タイトル
s1 = deck.add_slide(template="title")
s1.title = "NIR-HSIを用いた毛髪診断システムの開発"
s1.author = "データサイエンスでひも解く漢方と毛髪の関係<br>礒辺真人"
s1.show()
```

```python
# スライド2: 目次
s2 = deck.add_slide(template="agenda")
s2.title = "目次"
s2.markdown = """
1. 背景
2. 漢方体質アンケート
3. アンケート結果解析
4. 毛髪と漢方体質
5. 今後の展望と課題
"""
s2.show()
```

```python
# スライド3: 背景
s3 = deck.add_slide(title="背景：漢方体質とは？気血水の基本概念")
s3.markdown = "漢方医学では、人体は**「気」「血」「水」**の3要素がバランスよく巡ることで健康が保たれると考えられている。"

g = s3.grid(col="3")
g.left.markdown = """::: card(yellow)
**気：生命エネルギー・代謝**
自律神経系や代謝を司る活力
- **気虚**: エネルギー不足・疲労・胃腸虚弱
- **気滞**: 気の滞り・抑うつ・膨満感
- **気逆**: 気の逆流・のぼせ・動悸
:::"""

g.center.markdown = """::: card(red)
**血：血液・ホルモン・栄養**
全身に酸素や栄養を届ける働き
- **血虚**: 栄養不足・肌荒れ・貧血
- **瘀血**: 血行不良・冷えのぼせ
:::"""

g.right.markdown = """::: card(blue)
**水：体液・リンパ・水分代謝**
血液以外の水分や分泌液
- **水滞 / 水毒**: 水分代謝異常・むくみ・めまい・頭重感
:::"""

s3.markdown = """::: memo
**日本独自の発展（古方派）**：中国の思弁的な理論に対し、日本の江戸時代（吉益東洞ら）の実証的な漢方医学において、病態をシンプルに捉える指標として発展・体系化。
:::"""

s3.show()
```

```python
# スライド4: アンケート解析 (Plotlyグラフ埋め込み)
s4 = deck.add_slide(title="アンケート解析 (ダミーデータ)")
s4.markdown = "漢方**体質**と呼んでいるが、実際に体質と呼べるような分類があるのか？\n各ステージごとに、GMM→BIC/AICを実施。"

# ダミーのPlotlyグラフを作成
import numpy as np
x = np.arange(1, 10)
y_bic = np.random.randn(9).cumsum()
y_aic = np.random.randn(9).cumsum()
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y_bic, mode='lines+markers', name='BIC'))
fig.add_trace(go.Scatter(x=x, y=y_aic, mode='lines+markers', name='AIC'))
fig.update_layout(title="BIC/AIC グラフ", height=400, margin=dict(l=0,r=0,t=40,b=0))

g = s4.grid(col="2:1")
g.left.markdown = """* **GMM (混合ガウスモデル)**:
データを確率分布の重なりとして捉える手法。
* **BIC/AIC（ベイズ/赤池情報量規準）**:
「データの当てはまり」と「シンプルさ」の**バランスを評価する統計指標**。"""

g.right.chart = fig

s4.show()
```

```python
# スライド5: 3Dデータ分析
s5 = deck.add_slide(title="3Dデータ分析結果 (本編)")
s5.markdown = "多次元データの可視化（マウス操作で360度回転・ズーム・ホバーが可能）"

# 3Dグラフ
df = px.data.iris()
fig3d = px.scatter_3d(df, x='sepal_length', y='sepal_width', z='petal_width',
              color='species')
fig3d.update_layout(height=450, margin=dict(l=0,r=0,t=0,b=0))

s5.chart = fig3d
s5.show()
```
